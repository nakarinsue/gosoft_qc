```python
import os
os.environ["NLS_LANG"] = "THAI_THAILAND.AL32UTF8"  # ตั้งค่า Oracle NLS

import oracledb
import pandas as pd
from IPython.display import display
import ipywidgets as widgets
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side

# pandas display config
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.unicode.east_asian_width', True)


def show_with_copy(df, table_name):
    """แสดง DataFrame พร้อมปุ่ม Copy"""
    display(df)
    button = widgets.Button(description=f"📋 Copy {table_name}")
    output = widgets.Output()

    def on_copy(b):
        df.to_clipboard(index=False)
        with output:
            output.clear_output()
            print(f"✅ คัดลอก {table_name} ไป clipboard แล้ว (พร้อม header)")

    button.on_click(on_copy)
    display(button, output)


def compare_table_between_db(db1, db2, table_name, values, service=None, schema="ONLSTD"):
    """
    เปรียบเทียบข้อมูลจาก 2 DB (db1, db2) table เดียวกัน
    """
    df1 = db1.fetch_tables(values, service, [table_name])["results"].get(f"{schema}.{table_name}")
    df2 = db2.fetch_tables(values, service, [table_name])["results"].get(f"{schema}.{table_name}")

    if df1 is None or df2 is None:
        print(f"⚠️ {table_name}: ข้อมูลไม่ครบ (DB1={df1 is not None}, DB2={df2 is not None})")
        return None

    if df1.shape != df2.shape:
        print(f"❌ {table_name}: shape ต่างกัน {df1.shape} vs {df2.shape}")
        return None

    diff = df1.compare(df2, keep_shape=True, keep_equal=False)
    if diff.empty:
        print(f"✅ {table_name}: ข้อมูลเหมือนกันทั้งหมด")
        return True
    else:
        print(f"⚠️ {table_name}: พบความต่าง")
        display(diff)
        return diff


class OracleDB:
    def __init__(self, user, password, dsn):
        self.user = user
        self.password = password
        self.dsn = dsn
        self.conn = None

    def connect(self):
        """เชื่อมต่อ Oracle DB ด้วย UTF-8 รองรับภาษาไทย"""
        try:
            self.conn = oracledb.connect(
                user=self.user,
                password=self.password,
                dsn=self.dsn,
                encoding="UTF-8",
                nencoding="UTF-8"
            )
            print("✅ Connected to Oracle Database (UTF-8)")
        except Exception as e:
            print("❌ Connection failed:", e)

    def set_values_config(self, values: str, service: str = None):
        """สร้าง WHERE condition จากค่าที่ส่งมา"""
        values = str(values).strip()
        if len(values) < 6:
            flag_code = False
            strvalue = f"WHERE VENDOR_CODE = '{values}'"
            if service:
                strvalue += f" AND SERVICE_ID = '{service}'"
            value_data = {"VENDOR_CODE": values, "SERVICE_ID": service, "VENDOR_ID": None}
        else:
            flag_code = True
            strvalue = f"WHERE VENDOR_ID = '{values}'"
            if service:
                strvalue += f" AND SERVICE_ID = '{service}'"
            value_data = {"VENDOR_CODE": None, "SERVICE_ID": service, "VENDOR_ID": values}
        return {"flag_code": flag_code, "strvalue": strvalue, "value_data": value_data}

    def run_query(self, query, fetch=True, show=True, title="QueryResult"):
        """รัน query (SELECT / UPDATE)"""
        try:
            with self.conn.cursor() as cur:
                cur.execute(query)
                if fetch:
                    cols = [c[0] for c in cur.description]
                    rows = cur.fetchall()
                    # แปลงแต่ละค่าเป็น str เพื่อรองรับ Unicode
                    rows_clean = [[str(v) if v is not None else "" for v in r] for r in rows]
                    df = pd.DataFrame(rows_clean, columns=cols)
                    if show:
                        show_with_copy(df, title)
                    return df
                else:
                    self.conn.commit()
                    print(f"✅ Executed: {query[:80]}...")
                    return None
        except Exception as e:
            print(f"❌ Error: {e}")
            return pd.DataFrame({"Error": [str(e)], "Query": [query]}) if fetch else None

    def fetch_tables(self, values: str, service: str = None, tables: list = None, export_excel: str = None):
        """
        ดึงข้อมูลจากหลาย table + แสดงผลใน Jupyter
        export_excel: ถ้าใส่ชื่อไฟล์ เช่น "output.xlsx" → export เป็น Excel ด้วย
        """
        if not self.conn:
            print("ยังไม่ได้เชื่อมต่อ DB")
            return
        if not tables:
            print("กรุณาส่ง list ของชื่อตารางเข้ามา")
            return

        config = self.set_values_config(values, service)
        all_results = {}

        for table in tables:
            table_full = "ONLSTD." + table
            query = f"SELECT * FROM {table_full} {config['strvalue']}"

            try:
                cur = self.conn.cursor()
                cur.execute(query)
                cols = [c[0] for c in cur.description]
                rows = cur.fetchall()
                rows_clean = [[str(v) if v is not None else "" for v in r] for r in rows]
                df = pd.DataFrame(rows_clean, columns=cols)

            except Exception as e:
                print(f"❌ Error querying {table_full}:", e)
                all_results[table_full] = None
                continue

            if df.empty:
                print(f"⚠️ {table_full}: ไม่พบข้อมูล")
                all_results[table_full] = None
            else:
                print(f"\n✅ {table_full}: พบข้อมูล {len(df)} แถว")
                show_with_copy(df, table_full)
                all_results[table_full] = df

        # Export Excel ถ้าผู้ใช้ระบุ
        if export_excel:
            wb = Workbook()
            ws = wb.active
            ws.title = "Export"
            header_fill = PatternFill(start_color="92D050", end_color="92D050", fill_type="solid")
            column_fill = PatternFill(start_color="FFD966", end_color="FFD966", fill_type="solid")
            border = Border(left=Side(style="thin"), right=Side(style="thin"),
                            top=Side(style="thin"), bottom=Side(style="thin"))
            row = 1

            for table_name, df in all_results.items():
                ws.cell(row=row, column=1, value=table_name)
                ws.cell(row=row, column=1).fill = header_fill
                ws.cell(row=row, column=1).font = Font(bold=True)
                row += 2
                if df is not None and not df.empty:
                    for col_num, col_name in enumerate(df.columns, 1):
                        cell = ws.cell(row=row, column=col_num, value=col_name)
                        cell.fill = column_fill
                        cell.font = Font(bold=True)
                        cell.border = border
                        cell.alignment = Alignment(horizontal="center")
                    for r in df.itertuples(index=False):
                        row += 1
                        for col_num, val in enumerate(r, 1):
                            cell = ws.cell(row=row, column=col_num, value=val)
                            cell.border = border
                    row += 3
                else:
                    ws.cell(row=row, column=1, value="No Data")
                    row += 3

            # Auto adjust column width
            for col in ws.columns:
                max_length = 0
                col_letter = col[0].column_letter
                for cell in col:
                    if cell.value:
                        max_length = max(max_length, len(str(cell.value)))
                ws.column_dimensions[col_letter].width = max_length + 2

            wb.save(export_excel)
            print(f"Exported all results to {export_excel}")

        return {"flag_code": config["flag_code"], "value_data": config["value_data"], "results": all_results}
```

```python
import xml.etree.ElementTree as ET
from xml.dom import minidom

def dict_to_xml_nested(d, root_name=None):
    """
    แปลง dict nested เป็น XML
    """
    def build_element(parent, dict_data):
        for key, value in dict_data.items():
            if isinstance(value, dict):
                child = ET.SubElement(parent, key)
                build_element(child, value)
            else:
                child = ET.SubElement(parent, key)
                child.text = str(value)

    if root_name:
        root = ET.Element(root_name)
        build_element(root, d)
    else:
        # ถ้าไม่มี root_name ใช้ key แรกเป็น root
        first_key = list(d.keys())[0]
        root = ET.Element(first_key)
        build_element(root, d[first_key])

    xml_str = ET.tostring(root, encoding='utf-8')
    parsed = minidom.parseString(xml_str)
    return parsed.toprettyxml(indent="  ")
```

```python
def wrap_soap_envelope(xml_body_str, action="ACTION_NAME"):
    envelope = ET.Element("soapenv:Envelope", attrib={
        "xmlns:soapenv": "http://schemas.xmlsoap.org/soap/envelope/",
        "xmlns:por": "http://portal.cs/"
    })
    header = ET.SubElement(envelope, "soapenv:Header")
    body = ET.SubElement(envelope, "soapenv:Body")
    cs_service = ET.SubElement(body, "por:CSService")
    arg0 = ET.SubElement(cs_service, "arg0")
    
    # ใส่ CDATA ของ XML body
    arg0.text = f"<![CDATA[{xml_body_str}]]>"

    xml_str = ET.tostring(envelope, encoding='utf-8')
    parsed = minidom.parseString(xml_str)
    return parsed.toprettyxml(indent="  ")
```

```python
import os
import xml.etree.ElementTree as ET
from xml.dom import minidom
from copy import deepcopy
import requests

class SoapRequestManager:
    def __init__(self, soap_url=None, soap_action=None):
        self.before_request = []
        self.after_request = []
        self.last_dict = None
        self.last_xml = None
        self.last_soap_xml = None
        self.soap_url = soap_url
        self.soap_action = soap_action

    # -------------------
    # 1. generate dict
    # -------------------
    def generate_dict(self, **kwargs):
        cancel_id = "|".join([kwargs['store_id'], kwargs['pos_no'], kwargs['shift_no'],
                              "".join(kwargs['sys_date'].split("/")) + "".join(kwargs['sys_time'].split(":"))])
        data_dict = {
            "HQ_REQUEST": {
                "SERVICE_BOX": {
                    "CSSERVICE_BOX": {
                        "ADDRESS": {
                            "VENDOR_ID": kwargs['vendor_id'],
                            "SERVICE_ID": kwargs['service_id'],
                            "METHOD": "DataExchange"
                        },
                        "SYSTEM_DATA": {
                            "PROGRAM_VERSION": "7.00.1610.0000",
                            "PAYMENT_CHANNEL": f"C{kwargs['service_id']}",
                            "STORE_ID": kwargs['store_id'],
                            "POS_NO": kwargs['pos_no'],
                            "EMPLOYEE_ID": kwargs['employee_id'],
                            "SYS_DATE": kwargs['sys_date'],
                            "SYS_TIME": kwargs['sys_time'],
                            "BUS_DATE": kwargs['bus_date'],
                            "BUS_TIME": kwargs['bus_time'],
                            "SHIFT_NO": kwargs['shift_no'],
                            "POS_TAX_ID": kwargs['pos_tax_id'],
                            "ZONE": kwargs['zone'],
                            "COMMON_TRN_NO": "0",
                            "CANCEL_ID": cancel_id
                        },
                        "BILL": {
                            "TRANS_NO": "0",
                            "BILL_PAYMENT": {
                                "GROUP_NO": "1",
                                "IS_DEFAULT": "true",
                                "BILL_INFO": {
                                    "SEQ_NO": "0",
                                    "ITEM_SEQ_NO": "0",
                                    "COUNT_BIZ_SEQ": "0",
                                    "IS_DEFAULT": "true",
                                    "VENDOR_ID": kwargs['vendor_id'],
                                    "SERVICE_ID": kwargs['service_id'],
                                    "TRANS_TYPE": "N",
                                    "REC_TRANS_TYPE": "MB",
                                    "TXID": "|".join(kwargs['txid_list']),
                                    "AMOUNT": {
                                        "BILL_AMT": str(kwargs['bill_amt']),
                                        "ROUND_BILL_AMT": "0",
                                        "VAT_AMT": "0"
                                    }
                                }
                            }
                        }
                    }
                }
            }
        }

        self.last_dict = data_dict
        self.before_request.append(deepcopy(data_dict))
        return data_dict

    # -------------------
    # 2. dict → XML
    # -------------------
    def dict_to_xml(self, data_dict):
        def build_element(parent, dict_data):
            for key, value in dict_data.items():
                if isinstance(value, dict):
                    child = ET.SubElement(parent, key)
                    build_element(child, value)
                else:
                    child = ET.SubElement(parent, key)
                    child.text = str(value)

        first_key = list(data_dict.keys())[0]
        root = ET.Element(first_key)
        build_element(root, data_dict[first_key])

        xml_str = ET.tostring(root, encoding='utf-8')
        parsed = minidom.parseString(xml_str)
        xml_pretty = parsed.toprettyxml(indent="  ")
        self.last_xml = xml_pretty
        return xml_pretty

    # -------------------
    # 3. wrap SOAP
    # -------------------
    def wrap_soap(self, xml_body_str, action="ACTION"):
        envelope = ET.Element("soapenv:Envelope", attrib={
            "xmlns:soapenv": "http://schemas.xmlsoap.org/soap/envelope/",
            "xmlns:por": "http://portal.cs/"
        })
        ET.SubElement(envelope, "soapenv:Header")
        body = ET.SubElement(envelope, "soapenv:Body")
        cs_service = ET.SubElement(body, "por:CSService")
        arg0 = ET.SubElement(cs_service, "arg0")
        arg0.text = f"<![CDATA[{xml_body_str}]]>"

        xml_str = ET.tostring(envelope, encoding='utf-8')
        parsed = minidom.parseString(xml_str)
        soap_xml = parsed.toprettyxml(indent="  ")
        self.last_soap_xml = soap_xml
        return soap_xml

    # -------------------
    # 4. ส่ง SOAP + บันทึก response
    # -------------------
    def send_soap(self):
        if not self.soap_url:
            raise ValueError("SOAP URL ไม่ได้กำหนด")
        headers = {
            "Content-Type": "text/xml; charset=utf-8",
        }
        if self.soap_action:
            headers["SOAPAction"] = self.soap_action

        response = requests.post(self.soap_url, data=self.last_soap_xml.encode("utf-8"), headers=headers)
        self.after_request.append(response.text)
        return response.text

    # -------------------
    # 5. ส่ง SOAP แบบ batch + export XML
    # -------------------
    def process_batch(self, batch_data: list, folder: str = "soap_batch", send_request: bool = True):
        """
        batch_data: list ของ dict kwargs สำหรับแต่ละรายการ
        folder: โฟลเดอร์เก็บไฟล์ XML
        send_request: True → ส่ง SOAP, False → สร้าง XML เท่านั้น
        """
        import os
        os.makedirs(folder, exist_ok=True)
        for idx, data_kwargs in enumerate(batch_data):
            self.generate_dict(**data_kwargs)
            xml_body = self.dict_to_xml(self.last_dict)
            soap_xml = self.wrap_soap(xml_body)

            txid_str = "_".join(data_kwargs['txid_list'])
            filename = f"{idx+1}_{txid_str}.xml"
            filepath = os.path.join(folder, filename)
            with open(filepath, "w", encoding="utf-8") as f:
                f.write(soap_xml)
            print(f"✅ Exported XML: {filepath}")

            if send_request:
                response_text = self.send_soap()
                print(f"📩 Sent SOAP for TXID: {txid_str}, Response length: {len(response_text)}")
```

```python
import base64
import requests  # ใช้ส่ง SOAP จริง
import xml.etree.ElementTree as ET
import os
from copy import deepcopy

class SoapRequestManager:
    def __init__(self, soap_url=None, soap_action=None):
        self.soap_url = soap_url
        self.soap_action = soap_action
        self.last_xml = None
        self.last_soap_xml = None
        self.before_request = []
        self.after_request = []

    def dict_to_soap(self, data_dict, action=None):
        from xml.sax.saxutils import escape
        xml_body = self.dict_to_xml(data_dict)
        self.last_xml = xml_body
        soap_xml = f"""<soapenv:Envelope xmlns:soapenv="http://schemas.xmlsoap.org/soap/envelope/" xmlns:por="http://portal.cs/">
<soapenv:Header/>
<soapenv:Body>
<por:CSService>
<arg0><![CDATA[{xml_body}]]></arg0>
</por:CSService>
</soapenv:Body>
</soapenv:Envelope>"""
        self.last_soap_xml = soap_xml
        return soap_xml

    def dict_to_xml(self, data_dict):
        def build_elem(parent, d):
            for k, v in d.items():
                elem = ET.SubElement(parent, k)
                if isinstance(v, dict):
                    build_elem(elem, v)
                elif isinstance(v, list):
                    elem.text = "|".join(str(x) for x in v)
                else:
                    elem.text = str(v) if v is not None else ""
        root_key = list(data_dict.keys())[0]
        root_elem = ET.Element(root_key)
        build_elem(root_elem, data_dict[root_key])
        return ET.tostring(root_elem, encoding="unicode")

    def xml_to_dict(self, xml_str=None):
        if xml_str is None:
            xml_str = self.last_xml
            if xml_str is None:
                raise ValueError("No XML to parse")

        def parse_element(elem):
            children = list(elem)
            if not children:
                text = elem.text.strip() if elem.text else ""
                if "|" in text:
                    return text.split("|")
                return text
            else:
                result = {}
                for child in children:
                    result[child.tag] = parse_element(child)
                return result

        root = ET.fromstring(xml_str)
        return {root.tag: parse_element(root)}

    def save_before_request(self, data_dict):
        self.before_request.append(deepcopy(data_dict))

    def save_after_request(self, response_dict):
        self.after_request.append(deepcopy(response_dict))

    def process_batch(self, batch_dicts, folder="soap_xmls", send_request=False):
        os.makedirs(folder, exist_ok=True)
        for idx, data in enumerate(batch_dicts, 1):
            # save before request
            self.save_before_request(data)

            # สร้าง SOAP XML
            soap_xml = self.dict_to_soap({"HQ_REQUEST": data}, action="CSService")

            # export XML
            filename = os.path.join(folder, f"request_{idx}.xml")
            with open(filename, "w", encoding="utf-8") as f:
                f.write(soap_xml)

            response_dict = {}
            if send_request and self.soap_url:
                headers = {
                    "Content-Type": "text/xml;charset=UTF-8",
                }
                if self.soap_action:
                    headers["SOAPAction"] = self.soap_action
                try:
                    r = requests.post(self.soap_url, data=soap_xml.encode("utf-8"), headers=headers)
                    r.raise_for_status()
                    # assume response content is Base64 encoded string inside <arg0>
                    root = ET.fromstring(r.content)
                    arg0 = root.find(".//arg0")
                    if arg0 is not None and arg0.text:
                        decoded = base64.b64decode(arg0.text).decode("utf-8")
                        response_dict = {"decoded_response": decoded, "raw_base64": arg0.text}
                    else:
                        response_dict = {"decoded_response": None, "raw_base64": None}
                except Exception as e:
                    response_dict = {"error": str(e)}

            # save after request
            self.save_after_request(response_dict)
```

```python
import datetime
from xml.etree.ElementTree import Element, SubElement, tostring, fromstring
from xml.dom import minidom
import requests
import base64
import re
class HQRequestXML:
    def __init__(self, vendor_id, service_id, method):
        self.vendor_id = vendor_id
        self.service_id = service_id
        self.method = method
        self.data_fields = {}
        self.auto_fill_datetime()
        self.response_data = {}
        self.response_dict = {}

    # Auto fill date/time
    def auto_fill_datetime(self):
        now = datetime.datetime.now()
        self.data_fields['SYS_DATE'] = now.strftime("%Y/%m/%d")
        self.data_fields['SYS_TIME'] = now.strftime("%H:%M:%S")
        self.data_fields['BUS_DATE'] = now.strftime("%Y/%m/%d")
        self.data_fields['BUS_TIME'] = now.strftime("%H:%M:%S")

    # Set data field dynamic
    def set_data_field(self, key, value):
        self.data_fields[key] = value

    # Generate XML string
    def generate_xml(self):
        root = Element("HQ_REQUEST")
        service_box = SubElement(root, "SERVICE_BOX")
        
        # ADDRESS
        address = SubElement(service_box, "ADDRESS")
        SubElement(address, "VENDOR_ID").text = self.vendor_id
        SubElement(address, "SERVICE_ID").text = self.service_id
        SubElement(address, "METHOD").text = self.method
        
        # DATA
        data = SubElement(service_box, "DATA")
        for key, value in self.data_fields.items():
            SubElement(data, key).text = str(value) if value is not None else ""
        
        xml_str = minidom.parseString(tostring(root, 'utf-8')).toprettyxml(indent="    ")
        return xml_str

    # Save response XML
    def save_response(self, xml_response):
        try:
            root = fromstring(xml_response)
        except Exception as e:
            raise ValueError(f"Invalid XML format: {e}")

        self.response_data = {child.tag: child.text for child in root}
        self.response_dict = {}
        for k, v in self.response_data.items():
            if v is not None and '|' in v:
                self.response_dict[k] = v.split('|')
            else:
                self.response_dict[k] = v

    # Return processed response
    def get_response_dict(self):
        return self.response_dict
    def extract_and_decode_soap(soap_response: str) -> str:
        """
        ดึง Base64 จาก <return> ... </return> แล้ว decode เป็น XML string
        """
        # ดึงเนื้อหาใน <return> ... </return>
        match = re.search(r"<return>(.*?)</return>", soap_response, re.DOTALL)
        if not match:
	        raise ValueError("ไม่พบแท็ก <return> ใน SOAP response")

        base64_content = match.group(1).strip()
	
        # Decode Base64
        try:
        	decoded_xml = base64.b64decode(base64_content).decode('utf-8')
        except Exception as e:
	        raise ValueError(f"ไม่สามารถ decode Base64 ได้: {e}")

        return decoded_xml
    # POST XML to API with status check and Base64 decode
    def post_api(self, url, headers=None, timeout=30):
        xml_request = self.generate_xml()
        xml_body = f"""<soapenv:Envelope xmlns:soapenv="http://schemas.xmlsoap.org/soap/envelope/" xmlns:por="http://portal.cs/">
		<soapenv:Header/>
		<soapenv:Body>
		<por:CSService>
		<arg0><![CDATA[{xml_request}]]></arg0>
		</por:CSService>
		</soapenv:Body>
		</soapenv:Envelope>"""
        print(xml_body)
        if headers is None:
            headers = {'Content-Type': 'application/xml'}

        response = requests.post(url, data=xml_body, headers=headers, timeout=timeout)
        
        # ตรวจสอบ status code
        if response.status_code != 200:
            raise ConnectionError(f"Error: HTTP Status {response.status_code} - {response.text}")

        # Decode Base64
        resp_content = response.content
        try:
        	decoded_data = base64.b64decode(resp_content).decode('utf-8')
        except Exception:
	# fallback: assume it's plain XML
        	decoded_data = resp_content.decode('utf-8')

        # แปลง decoded string เป็น XML
        	self.save_response(decoded_data)
        return decoded_data


# =======================
# ตัวอย่างใช้งาน
# =======================
hq = HQRequestXML(vendor_id="0994000160151", service_id="16", method="DataExchange")
hq.set_data_field("PAYMENT_CHANNEL", "C05")
hq.set_data_field("STORE_ID", "09891")
hq.set_data_field("ITEM_NAME", "ชำระค่าขยะ")
hq.set_data_field("BILL_AMT", "240")
hq.set_data_field("EMPLOYEE_ID", "0999902")
hq.set_data_field("POS_TAX_ID", "3421234567890")

url = "http://qacspos-temp.counterservice.co.th:80/DCWSCDSONLINE/WSCDSService"

try:
    decoded_xml = hq.post_api(url)
    print("===== Decoded XML =====")
    print(decoded_xml)
    xml_result = hq.extract_and_decode_soap(decoded_xml)
    print(xml_result)
    print("===== Response Dict =====")
    print(hq.get_response_dict())
except Exception as e:
    print("Error:", e)
```
