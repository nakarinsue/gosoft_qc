```python
import os
import shutil

def copy_files(file_names, source_folder, dest_folder):
    """
    คัดลอกไฟล์จาก source_folder → dest_folder
    file_names: list ของชื่อไฟล์ที่ต้องการหา เช่น ['csaccbatuat01.34749832']
    """

    # สร้างโฟลเดอร์ปลายทางถ้ายังไม่มี
    os.makedirs(dest_folder, exist_ok=True)

    not_found = []  # เก็บไฟล์ที่หาไม่เจอ
    copied = []     # เก็บไฟล์ที่ถูก copy

    # loop หาไฟล์ที่ match ชื่อ
    for fname in file_names:
        found = False
        for file in os.listdir(source_folder):
            if fname in file:  # ถ้าเจอชื่อที่ match
                src = os.path.join(source_folder, file)
                dst = os.path.join(dest_folder, file)
                shutil.copy2(src, dst)
                copied.append(file)
                found = True
        if not found:
            not_found.append(fname)

    return copied, not_found
```
